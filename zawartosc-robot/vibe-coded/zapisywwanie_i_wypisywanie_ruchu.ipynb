```python
# Import some matplolib shortcuts for Jupyter notebook
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
from pypot.creatures import PoppyTorso
from pypot.primitive.move import MoveRecorder
import time
from copy import deepcopy
from pypot.primitive.move import MovePlayer
import keyboard
import threading

poppy = PoppyTorso(camera='dummy')

starting_pos = {
    'head_z': 0,
    'head_y': -60,
    'l_shoulder_y': 0,
    'l_shoulder_x': 90,
    'l_elbow_y': 90,
    'r_shoulder_y': 0,
    'r_shoulder_x': -45,
    'r_elbow_y': -90
}
```

```python
if __name__ == '__main__':
       
    print(poppy.alias)
    recorder = MoveRecorder(poppy, 50, poppy.motors)

    print(poppy.motors)

    for i in range(5):
        time.sleep(1)
        print("starting in: ", i)

    # global flag
    stop_flag = False

    # wait for flag
    def listen_for_exit():
        global stop_flag
        input("Press 'x' and hit Enter to stop recording...\n")  
        stop_flag = True # set flag

    # start thread
    listener_thread = threading.Thread(target=listen_for_exit)
    listener_thread.start()

    # record
    while not stop_flag:
        print("Recording started...")
        recorder.start()

        for time_passed in range(50):  # 50 * 1s = 50s total
            if stop_flag:
                break
            
            
            ##==============vibe coding===============
            
            #wypisywanie aktualnych pozycji silników
            print(f"\nTime: {time_passed}s")
            for m in poppy.motors:
                print(f"{m.name}: {m.present_position:.2f}")
                
            time.sleep(1)

        recorder.stop()
        print("Recording stopped.")

        if stop_flag:
            break

        print("Restarting recording...")
        time.sleep(1) 

    print("\nStopping recorder...")
    recorder.stop()

    my_first_move = deepcopy(recorder.move)
    player = MovePlayer(poppy, my_first_move)
    player.start()
    time.sleep(time_passed+1)
    print("end")
    
    with open('my-first-demo.move', 'w') as f:
        my_first_move.save(f)
    
    for m in poppy.motors:
        m.compliant = True
```
