```python
import tkinter as tk
import time

# Ustawienia główne okna
root = tk.Tk()
root.title("Mrugające oczy")

# Ustawienia okna
canvas_width, canvas_height = 800, 480
canvas = tk.Canvas(root, width=canvas_width, height=canvas_height, bg="white")
canvas.pack()

# Ustawienia oczu
eye_radius = 40
eye_y = canvas_height // 2
left_eye_x = canvas_width // 3
right_eye_x = 2 * canvas_width // 3

# Zmienna przechowująca stan mrugnięcia
is_blinking = False

# Funkcja rysująca oczy
def draw_eyes():
    # Usunięcie poprzednich elementów
    canvas.delete("all")
    
    # Rysowanie lewego oka
    if not is_blinking:
        canvas.create_oval(left_eye_x - eye_radius, eye_y - eye_radius, left_eye_x + eye_radius, eye_y + eye_radius, fill="grey")
        canvas.create_oval(left_eye_x - eye_radius // 3, eye_y - eye_radius // 3, left_eye_x + eye_radius // 3, eye_y + eye_radius // 3, fill="black")
    else:
        canvas.create_rectangle(left_eye_x - eye_radius, eye_y - eye_radius // 4, left_eye_x + eye_radius, eye_y + eye_radius // 4, fill="grey")

    # Rysowanie prawego oka
    if not is_blinking:
        canvas.create_oval(right_eye_x - eye_radius, eye_y - eye_radius, right_eye_x + eye_radius, eye_y + eye_radius, fill="grey")
        canvas.create_oval(right_eye_x - eye_radius // 3, eye_y - eye_radius // 3, right_eye_x + eye_radius // 3, eye_y + eye_radius // 3, fill="black")
    else:
        canvas.create_rectangle(right_eye_x - eye_radius, eye_y - eye_radius // 4, right_eye_x + eye_radius, eye_y + eye_radius // 4, fill="grey")

# Funkcja zmieniająca stan mrugnięcia
def blink():
    global is_blinking
    is_blinking = True
    draw_eyes()
    root.after(100, end_blink)  # Czas trwania mrugnięcia (100 ms)

# Funkcja kończąca mrugnięcie
def end_blink():
    global is_blinking
    is_blinking = False
    draw_eyes()
    # Ustawienie kolejnego mrugnięcia po losowym czasie
    root.after(2000, blink)  # Mrugnięcie co 2 sekundy

# Uruchomienie pierwszego rysowania i mrugnięcia
draw_eyes()
root.after(2000, blink)  # Pierwsze mrugnięcie po 2 sekundach

# Start pętli głównej Tkinter
root.mainloop()
```
