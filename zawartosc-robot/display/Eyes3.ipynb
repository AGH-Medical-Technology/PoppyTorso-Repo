```python
import tkinter as tk
import sys
import os

if os.environ.get('DISPLAY','') == '':
    print('no display, using 0.0')
    os.environ.__setitem__('DISPLAY',':0')
# Ustawienia główne okna
root = tk.Tk()
root.title("Mrugające oczy")

# Ustawienia okna
canvas_width, canvas_height = 800, 600
canvas = tk.Canvas(root, width=canvas_width, height=canvas_height, bg="black")
canvas.pack()

# Ustawienia oczu
eye_width = 60   # Szerokość oka
eye_height = 80  # Wysokość oka
eye_y = canvas_height // 3
left_eye_x = canvas_width // 3
right_eye_x = 2 * canvas_width // 3
pupil_offset = 30  # Przesunięcie źrenicy w kierunku krawędzi
pupil_size = eye_width // 2  # Rozmiar źrenicy

# Zmienna przechowująca stan oczu
is_blinking = False
look_direction = "center"

# Funkcja rysująca oczy
def draw_eyes():
    # Usunięcie poprzednich elementów
    canvas.delete("all")
    
    # Rysowanie lewego oka
    if not is_blinking:
        canvas.create_oval(left_eye_x - eye_width, eye_y - eye_height, left_eye_x + eye_width, eye_y + eye_height, outline="blue", width=2)
        if look_direction == "left":
            canvas.create_oval(left_eye_x - pupil_size // 2 - pupil_offset, eye_y - pupil_size // 2,
                               left_eye_x + pupil_size // 2 - pupil_offset, eye_y + pupil_size // 2, fill="blue")
        elif look_direction == "right":
            canvas.create_oval(left_eye_x - pupil_size // 2 + pupil_offset, eye_y - pupil_size // 2,
                               left_eye_x + pupil_size // 2 + pupil_offset, eye_y + pupil_size // 2, fill="blue")
        else:
            canvas.create_oval(left_eye_x - pupil_size // 2, eye_y - pupil_size // 2,
                               left_eye_x + pupil_size // 2, eye_y + pupil_size // 2, fill="blue")
    else:
        canvas.create_rectangle(left_eye_x - eye_width, eye_y - eye_height // 8, left_eye_x + eye_width, eye_y + eye_height // 8, fill="blue")

    # Rysowanie prawego oka
    if not is_blinking:
        canvas.create_oval(right_eye_x - eye_width, eye_y - eye_height, right_eye_x + eye_width, eye_y + eye_height, outline="blue", width=2)
        if look_direction == "left":
            canvas.create_oval(right_eye_x - pupil_size // 2 - pupil_offset, eye_y - pupil_size // 2,
                               right_eye_x + pupil_size // 2 - pupil_offset, eye_y + pupil_size // 2, fill="blue")
        elif look_direction == "right":
            canvas.create_oval(right_eye_x - pupil_size // 2 + pupil_offset, eye_y - pupil_size // 2,
                               right_eye_x + pupil_size // 2 + pupil_offset, eye_y + pupil_size // 2, fill="blue")
        else:
            canvas.create_oval(right_eye_x - pupil_size // 2, eye_y - pupil_size // 2,
                               right_eye_x + pupil_size // 2, eye_y + pupil_size // 2, fill="blue")
    else:
        canvas.create_rectangle(right_eye_x - eye_width, eye_y - eye_height // 8, right_eye_x + eye_width, eye_y + eye_height // 8, fill="blue")

# Funkcja zmieniająca stan mrugnięcia
def blink():
    global is_blinking
    is_blinking = True
    draw_eyes()
    root.after(100, end_blink)  # Czas trwania mrugnięcia (100 ms)

# Funkcja kończąca mrugnięcie
def end_blink():
    global is_blinking
    is_blinking = False
    draw_eyes()
    # Ustawienie kolejnego mrugnięcia po 5 sekundach
    root.after(5000, blink)  # Mrugnięcie co 5 sekund

# Funkcja zmieniająca kierunek patrzenia
def look_left():
    global look_direction
    look_direction = "left"
    draw_eyes()
    root.after(1000, look_right)  # Zmiana na patrzenie w prawo po 1 sekundzie

def look_right():
    global look_direction
    look_direction = "right"
    draw_eyes()
    root.after(1000, look_center)  # Powrót do centrum po 1 sekundzie

def look_center():
    global look_direction
    look_direction = "center"
    draw_eyes()
    # Uruchamiamy ruch oczu co 5 sekund, między mrugnięciami
    root.after(5000, look_left)

# Uruchomienie pierwszego rysowania, mrugnięcia i ruchu oczu
draw_eyes()
root.after(5000, blink)  # Pierwsze mrugnięcie po 5 sekundach
root.after(5000, look_left)  # Pierwszy ruch oczu po 5 sekundach

# Start pętli głównej Tkinter
root.mainloop()
```
