```python
!pip install pygame
import pygame
import sys
import time

# Inicjalizacja Pygame
pygame.init()

# Ustawienia ekranu
screen_width, screen_height = 800, 480  # Dopasuj rozdzielczość do ekranu
screen = pygame.display.set_mode((screen_width, screen_height))
pygame.display.set_caption("Mrugające oczy")

# Kolory
white = (255, 255, 255)
black = (0, 0, 0)
eye_color = (50, 50, 50)
blink_color = white

# Pozycje i rozmiary oczu
eye_radius = 40
eye_y = screen_height // 2
left_eye_x = screen_width // 3
right_eye_x = 2 * screen_width // 3

# Czas mrugnięcia
blink_interval = 2  # Sekundy między mrugnięciami
blink_duration = 0.1  # Czas trwania mrugnięcia

# Zmienna kontrolująca stan mrugnięcia
is_blinking = False
last_blink_time = time.time()

# Funkcja rysująca oczy
def draw_eyes():
    screen.fill(white)

    # Rysowanie lewego oka
    if not is_blinking:
        pygame.draw.circle(screen, eye_color, (left_eye_x, eye_y), eye_radius)  # Gałka oczna
        pygame.draw.circle(screen, black, (left_eye_x, eye_y), eye_radius // 3)  # Źrenica
    else:
        pygame.draw.rect(screen, blink_color, (left_eye_x - eye_radius, eye_y - eye_radius // 4, eye_radius * 2, eye_radius // 2))  # Zamknięte oko

    # Rysowanie prawego oka
    if not is_blinking:
        pygame.draw.circle(screen, eye_color, (right_eye_x, eye_y), eye_radius)  # Gałka oczna
        pygame.draw.circle(screen, black, (right_eye_x, eye_y), eye_radius // 3)  # Źrenica
    else:
        pygame.draw.rect(screen, blink_color, (right_eye_x - eye_radius, eye_y - eye_radius // 4, eye_radius * 2, eye_radius // 2))  # Zamknięte oko

    pygame.display.flip()

# Pętla główna
while True:
    # Obsługa zdarzeń
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            pygame.quit()
            sys.exit()

    # Sprawdzenie, czy powinno nastąpić mrugnięcie
    current_time = time.time()
    if current_time - last_blink_time >= blink_interval:
        is_blinking = True
        draw_eyes()
        time.sleep(blink_duration)  # Czas zamknięcia oczu
        is_blinking = False
        last_blink_time = current_time

    # Rysowanie otwartych oczu
    draw_eyes()

    # Ograniczenie liczby klatek na sekundę
    pygame.time.Clock().tick(30)
```
