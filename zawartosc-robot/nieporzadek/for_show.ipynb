```python
if __name__ == '__main__':
       
    print(poppy.alias)
    recorder = MoveRecorder(poppy, 50, poppy.motors)

    print(poppy.motors)

    for i in range(5):
        time.sleep(1)
        print("starting in: ", i)

    # global flag
    stop_flag = False

    # wait for flag
    def listen_for_exit():
        global stop_flag
        input("Press 'x' and hit Enter to stop recording...\n")  
        stop_flag = True # set flag

    # start thread
    listener_thread = threading.Thread(target=listen_for_exit)
    listener_thread.start()

    # record
    while not stop_flag:
        print("Recording started...")
        recorder.start()

        for time_passed in range(50):  # 50 * 1s = 50s total
            if stop_flag:
                break
            
            
            ##==============vibe coding===============
            
            #wypisywanie aktualnych pozycji silników
            print(f"\nTime: {time_passed}s")
            for m in poppy.motors:
                print(f"{m.name}: {m.present_position:.2f}")
                
            time.sleep(1)

        recorder.stop()
        print("Recording stopped.")

        if stop_flag:
            break

        print("Restarting recording...")
        time.sleep(1) 

    print("\nStopping recorder...")
    recorder.stop()

    my_first_move = deepcopy(recorder.move)
    player = MovePlayer(poppy, my_first_move, omit_morors = ['head_y'])
    player.start()
    time.sleep(time_passed+1)
    print("end")
    
    with open('my-first-demo.move', 'w') as f:
        my_first_move.save(f)
    
    for m in poppy.motors:
        m.compliant = True
```

### EVE eyes - komórke należy ręcznie zinteraptować 
Jak? - na górze obok run jest kwadracik "Interupt the kernel"

```python
import tkinter as tk
import itertools
import sys
import os

if os.environ.get('DISPLAY','') == '':
    print('no display, using 0.0')
    os.environ.__setitem__('DISPLAY',':0')
    
# Create the main window
root = tk.Tk()
root.title("EVE Gradient Eyes")

# Set up canvas size
canvas_width = 800
canvas_height = 600
root.geometry(f"{canvas_width}x{canvas_height}")
canvas = tk.Canvas(root, width=canvas_width, height=canvas_height, bg="black")
canvas.pack()

# Eye parameters
eye_radius = 100
animation_id = None

def shutdown():
    global animation_id
    if animation_id is not None:
        root.after_cancel(animation_id)
    root.quit()
    root.destroy()

# Function to draw gradient eye
def draw_gradient_eye(center_x, center_y, radius, color_start, color_end, layers=10):
    r_start, g_start, b_start = color_start
    r_end, g_end, b_end = color_end
    for i in range(layers):
        ratio = i / layers
        r = int(r_start + (r_end - r_start) * ratio)
        g = int(g_start + (g_end - g_start) * ratio)
        b = int(b_start + (b_end - b_start) * ratio)
        color = f'#{r:02x}{g:02x}{b:02x}'

        current_radius = radius * (1 - i / (layers * 2))
        canvas.create_oval(center_x - current_radius, center_y - current_radius,
                           center_x + current_radius, center_y + current_radius,
                           fill=color, outline="", tags="eye")

# Function to dynamically position and draw the eyes
def draw_eyes():
    canvas.delete("eye")  # Clear previous eyes
    # Calculate eye positions based on canvas size
    eye_left_x, eye_left_y = canvas_width * 0.3, canvas_height / 2.5
    eye_right_x, eye_right_y = canvas_width * 0.7, canvas_height / 2.5

    # Draw the left and right eyes with gradient effect
    draw_gradient_eye(eye_left_x, eye_left_y, eye_radius, (30, 144, 255), (0, 0, 0), layers=50)
    draw_gradient_eye(eye_right_x, eye_right_y, eye_radius, (30, 144, 255), (0, 0, 0), layers=50)

# # Animate the gradient glow
# def animate_glow():
#     pulse = itertools.cycle(range(-2, 3))  # Cycle through small radius changes

#     def pulse_eyes():
#         offset = next(pulse)
#         canvas.delete("eye")  # Clear previous eyes
#         draw_eyes()  # Redraw eyes
#         draw_gradient_eye(canvas_width * 0.3, canvas_height / 2.5, eye_radius + offset, (30, 144, 255), (0, 0, 0), layers=20)
#         draw_gradient_eye(canvas_width * 0.7, canvas_height / 2.5, eye_radius + offset, (30, 144, 255), (0, 0, 0), layers=20)
#         root.after(100, pulse_eyes)

#     pulse_eyes()
    
def animate_glow():
    global animation_id
    pulse = itertools.cycle(range(-2, 3))

    def pulse_eyes():
        global animation_id
        offset = next(pulse)
        canvas.delete("eye")
        draw_eyes()
        draw_gradient_eye(canvas_width * 0.3, canvas_height / 2.5, eye_radius + offset, (30, 144, 255), (0, 0, 0), layers=20)
        draw_gradient_eye(canvas_width * 0.7, canvas_height / 2.5, eye_radius + offset, (30, 144, 255), (0, 0, 0), layers=20)
        animation_id = root.after(100, pulse_eyes)

    pulse_eyes()
#     root.after(5000, shutdown)
#     root.bind("<Escape>", lambda e: shutdown())

# Draw initial eyes and start the glow animation
draw_eyes()
animate_glow()

root.mainloop()
```
