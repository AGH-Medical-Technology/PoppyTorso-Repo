```python
# Import some matplolib shortcuts for Jupyter notebook
%matplotlib inline
import time
import numpy as np
import matplotlib.pyplot as plt

from pypot.creatures import PoppyTorso

# if your Torso has no camera you should create your robot with:
poppy = PoppyTorso(camera='dummy')

starting_pos = {
    #'head_z': 0,
    #'head_y': 20,
    'l_shoulder_y': -93,
    'l_shoulder_x': 4.85,
    'l_elbow_y': -1.9,
    'r_shoulder_y': -91.36,
    'r_shoulder_x': -45,
    'r_elbow_y': -5.73
}

for m in poppy.motors:
    m.compliant = False
    
for motor_name, position in starting_pos.items():
    motor = getattr(poppy, motor_name)
    time.sleep(0.2)
    motor.goto_position(position, 0.2, wait=True)
```

```python
From the motor object you can access its registers. The main ones are:

**present_position**: the current position of the motor in degrees
**present_speed**: the current speed of the motor in degrees per second 
**present_load**: the current workload of the motor (in percentage of max load)
**present_temperature**: the temperature of the motor in celsius
**angle_limit**: the reachable limits of the motor (in degrees)
They can be accessed directly:
```

```python
#Basic 

poppy.r_shoulder_x.compliant = False

for m in poppy.motors:
    m.moving_speed = 100.0
    
poppy.r_shoulder_x.goal_position = -100
time.sleep(1)
print(poppy.r_shoulder_x.present_position)

poppy.r_shoulder_x.goal_position = -50
time.sleep(1)
print(poppy.r_shoulder_x.present_position)
```

```python
poppy.r_shoulder_x.goto_position(-100, 0.5, wait=True)
print(poppy.r_shoulder_x.present_position)
time.sleep(1)
poppy.r_shoulder_x.goto_position(-50, 0.5, wait=True)
print(poppy.r_shoulder_x.present_position)
```
